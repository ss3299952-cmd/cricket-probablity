```python
import os
import json
import pandas as pd

def load_match(path):
    with open(path) as f:
        match = json.load(f)
        info = match["info"]
        toss = info.get("toss", {})
        outcome = info.get("outcome", {})
        dates = info.get("dates", [None])
        base = {
            "match_id": os.path.basename(path).replace(".json", ""),
            "season": info.get("season"),
            "date": dates[0],
            "venue": info.get("venue"),
            "toss_winner": toss.get("winner"),
            "toss_decision": toss.get("decision"),
            "winner": outcome.get("winner"),
            "result": outcome.get("result"),
        }

    rows = []
    for inn_no, inn in enumerate(match.get("innings", []), start=1):
        if inn.get("super_over"):
            continue

        target = inn.get("target", {}).get("runs")
        for over in inn["overs"]:
            for ball in over["deliveries"]:
                wk = ball.get("wickets", [])
                row = dict(base)
                row.update({
                    "innings": inn_no,
                    "team": inn["team"],
                    "target_runs": target,
                    "over": over["over"],
                    "batter": ball["batter"],
                    "bowler": ball["bowler"],
                    "runs_batter": ball["runs"]["batter"],
                    "runs_extra": ball["runs"]["extras"],
                    "runs_total": ball["runs"]["total"],
                    "wickets": len(wk),
                    "player_out": wk[0]["player_out"] if wk else None,
                    "wicket_kind": wk[0]["kind"] if wk else None,
                })
                rows.append(row)

    return pd.DataFrame(rows)
```

```python
frames = []
bad = []
for f in files:
  try:        frames.append(load_match(os.path.join(folder, f)))
  except Exception as e:
    bad.append((f, str(e)))
  all_df = pd.concat(frames, ignore_index=True)
  print(all_df.shape)
  print(len(bad))
```
